```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = '/content/part-00000-0dd5f001-57d7-4ec1-a3f0-53ea28bba9c3-c000.csv.gz'

print("Loading Data (skipping bad lines)...")
df = pd.read_csv(file_path, compression='gzip', on_bad_lines='skip')

print("\n--- Basic Info ---")
print(df.info())

print("\n--- Platform Distribution (Play Count) ---")
platform_counts = df['platform'].value_counts()
display(platform_counts)

print("\n--- Average Play Duration by Platform (Seconds) ---")
platform_duration = df.groupby('platform')['play_duration'].mean().sort_values(ascending=False)
display(platform_duration)

print("\n--- Top 5 Watched Content Titles ---")
display(df['title'].value_counts().head(5))

sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='platform', order=df['platform'].value_counts().index, palette='viridis')
plt.title('Platform Popularity by Number of Plays')
plt.ylabel('Number of Plays')
plt.xlabel('Platform')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='platform', y='play_duration', order=platform_duration.index, palette='magma', errorbar=None)
plt.title('Average Play Duration by Platform (Seconds)')
plt.ylabel('Average Play Duration (s)')
plt.xlabel('Platform')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
